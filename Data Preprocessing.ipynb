```python
import pandas as pd
```

```python
#step 01:upload file
file_path=r"D:\Summer\House Price Prediction\Raw data\cleaned_house_price_data.csv"
df=pd.read_csv(file_path,encoding='latin1')
#view first 5 rows of dataset
print(df.head())
#view the statistics summary of dataset
print("Statistical Summary of Dataset")
print(df.describe())
print(df.dtypes)
```

Step 02: Separate target and features

```python
X=df.drop("price",axis=1)
Y=df["price"]

print("Feature Shape",X.shape)
print("Target shape",Y.shape)
```

Step 03: Identify categorical columns

```python
categorical_col=X.select_dtypes(include=['object']).columns
print("Categorical Columns:")
print(categorical_col.tolist())
```

Step 4 — Unique values check

```python
for col in categorical_col:
    print(f"Unique values for column '{col}':")
    print(X[col].nunique())
    print("\n")
```

```python
# Remove high-cardinality location column
X = X.drop("location", axis=1)

# Remove Price_per_Marla if it exists
if "Price_per_Marla" in X.columns:
    X = X.drop("Price_per_Marla", axis=1)

print("New feature shape:", X.shape)
```

Step 6 — One-Hot Encoding

```python
categorical_features = X.select_dtypes(include=['object']).columns
print(categorical_features.tolist())

X=pd.get_dummies(X, columns=categorical_features, drop_first=True)
print("New feature shape after one-hot encoding:", X.shape)
```

Step 7 — Final data types check

```python
print(X.dtypes.value_counts())
print("Object Columns:",X.select_dtypes(include=['object']).columns.tolist())
```

Step 8 — Train/Test Split

```python
import numpy as np
import sklearn

print("NumPy:", np.__version__)
print("Scikit-learn:", sklearn.__version__)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
print("X_train shape:",X_train.shape)
print("X_test shape:",X_test.shape)
print("Y_train shape:",Y_train.shape)
print("Y_test shape:",Y_test.shape)
```

Step 9 — Feature Scaling

```python
#print(X_train.describe().T)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit scaler on training data and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
```

```python
import joblib

joblib.dump(scaler, "scaler.pkl")

print("Scaler saved successfully!")
```

```python
# import pandas as pd

# X_train.to_csv("X_train.csv", index=False)
# X_test.to_csv("X_test.csv", index=False)

# Y_train.to_csv("y_train.csv", index=False)
# Y_test.to_csv("y_test.csv", index=False)

# print("Preprocessed data saved successfully!")

# Convert scaled arrays back to DataFrames
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)

# Save preprocessed data
X_train_scaled.to_csv("X_train_scaled.csv", index=False)
X_test_scaled.to_csv("X_test_scaled.csv", index=False)

Y_train.to_csv("y_train.csv", index=False)
Y_test.to_csv("y_test.csv", index=False)

print("Preprocessed data saved successfully!")
```
