```python
import pandas as pd
```

#step 01:upload file


```python
file_path=r"D:\Summer\House Price Prediction\data\Uncleaned House data.csv.csv"
df=pd.read_csv(file_path,encoding='latin1')
```

```python
#view first 5 rows of dataset
print(df.head())
```

```python
#view the statistics summary of dataset
print("Statistical Summary of Dataset")
print(df.describe())
```

```python
#step 02: check for duplicate rows
print("Duplicate rows in the dataset")
print(df.duplicated().sum())
```

```python
#step 03: check missing values
print("Missing values in the dataset")
print(df.isnull().sum())
```

```python
#step 04: identify wrong/invalid values
print(df[["price","baths","bedrooms","Area Size"]].describe())
print("Check zero values")
print("Zero Price:",(df["price"]==0).sum())
print("Zero Baths:",(df["baths"]==0).sum())
print("Zero Bedrooms:",(df["bedrooms"]==0).sum())
print("Zero Area Size:",(df["Area Size"]==0).sum())
print("Check negative values")
print("Negative Price:",(df["price"]<0).sum())
print("Negative Baths:",(df["baths"]<0).sum())
print("Negative Bedrooms:",(df["bedrooms"]<0).sum())
print("Negative Area Size:",(df["Area Size"]<0).sum())
print("Check for outliers")
print("Bedrooms greater than 10")
print(df[df["bedrooms"]>10][["price","city","baths","bedrooms","Area Size"]])
print("Baths greater than 10")
print(df[df["baths"]>10][["price","city","baths","bedrooms","Area Size"]])
print("Area Size greater than 500")
print(df[df["Area Size"]>500][["price","city","baths","bedrooms","Area Size"]])
```

```python
#step 05: handle purpose column
print(df["purpose"].value_counts())
```

```python
#step 06: filter the purpose column to keep only "For Sale" properties
df=df[df["purpose"]=="For Sale"]
print(df["purpose"].value_counts())
#reset the index of the dataframe after filtering as some rows may remains still in the dataframe
df.reset_index(drop=True)
```

```python
#step 07: clean price column
df=df[df["price"]>0]
print((df["price"]==0).sum())   
df.reset_index(drop=True)
print(df["price"].describe())
print("Minimum Price:",df["price"].min())
#hnadle price 1 anamoly in the dataset
print("Price==1:",(df["price"]==1).sum())
print(df[df["price"] == 1][
    ["price", "property_type", "location", "city", "Area Size", "bedrooms", "baths"]
])
df = df[df["price"] > 1]
df = df.reset_index(drop=True)
print("Minimum Price:", df["price"].min())
print("Price = 1:", (df["price"] == 1).sum())
#after the above line of code running we see the minimum price were 3 that is also an anomaly
print(df[df["price"] < 1000][
    ["price", "property_type", "location", "city", "Area Size", "bedrooms", "baths"]
])
print("Prices below 1000:", (df["price"] < 1000).sum())
#then we checked how many houses prices is below 1000 and we found 8 houses with price below 1000 so we will remove them from the dataset
print(df[df["price"] < 1000][
    ["price", "property_type", "location", "city", "Area Size", "bedrooms", "baths"]
])
df = df[df["price"] >= 1000]
df = df.reset_index(drop=True)
#verify
print("Prices below 1000:", (df["price"] < 1000).sum())
print("Minimum Price:", df["price"].min())
print("Price cleaned successfully")
```

```python
#step 08:area columns
print(df[["area", "Area Size", "Area Type", "Area Category"]].head(10))
print(df[["area", "Area Size", "Area Type", "Area Category"]].info())
print(df["Area Type"].value_counts())
print(df["Area Category"].value_counts())
#in the dataset there is a category of marla and kanal that ml model is confused with this so we will change both into the same unit for model convenience
#conversion of kanal into marla
df["Area_Marla"] = df["Area Size"]

df.loc[df["Area Type"] == "Kanal", "Area_Marla"] = (
    df.loc[df["Area Type"] == "Kanal", "Area Size"] * 20
)
print(df[["area", "Area Size", "Area Type", "Area_Marla"]].head(10))
print("Now all the area in the same unit(marla)")
#we created a new column called Area_Marla and converted all the area into marla unit so that our model can understand the data better
```

```python
#step 09: check for missing values in bedrooms and baths columns
print(df[["bedrooms", "baths"]].isnull().sum())
print("Bedrooms = 0:", (df["bedrooms"] == 0).sum())
print("Baths = 0:", (df["baths"] == 0).sum())
print("\nBedrooms:")
print(df["bedrooms"].describe())
print("\nBaths:")
print(df["baths"].describe())
#handle missing values in bedrooms and baths columns
print("Bedrooms = 0:")
print(df[df["bedrooms"] == 0][
    ["property_type", "price", "city", "Area_Marla", "baths"]
].head(20))

print("\nBaths = 0:")
print(df[df["baths"] == 0][
    ["property_type", "price", "city", "Area_Marla", "bedrooms"]
].head(20))
#check baths that are greater than 20
print("\nBaths > 20:")
print(df[df["baths"] > 20][
    ["property_type", "price", "city", "Area_Marla", "bedrooms", "baths"]
])
#403 baths in a house is an outlier so we will remove it from the dataset
df = df[df["baths"] <= 20]
df = df.reset_index(drop=True)
print("Maximum baths:", df["baths"].max())
```

```python
#step 10:check property type column
print(df["property_type"].value_counts())
```

```python
#step 11: check city column
print(df["city"].value_counts())
print("\nMissing cities:", df["city"].isnull().sum())
```

```python
#step 12: check province column
print(df["province_name"].value_counts())
print("\nMissing provinces:", df["province_name"].isnull().sum())
```

```python
#step 13: check latitude or longitude column
print(df[["latitude", "longitude"]].isnull().sum())
print("\nLatitude:")
print(df["latitude"].describe())
print("\nLongitude:")
print(df["longitude"].describe())
print(df[df["latitude"] > 40][
    ["city", "latitude", "longitude", "price", "property_type"]
])
print(df[(df["longitude"] < 60) | (df["longitude"] > 80)][
    ["city", "latitude", "longitude", "price", "property_type"]
])
#remove the rows that have latitude greater than 40 and longitude less than 60 or greater than 80
df = df[df["latitude"] <= 40]
df = df[(df["longitude"] >= 60) & (df["longitude"] <= 80)]
df = df.reset_index(drop=True)
print("Remaining rows:", len(df))
print("Maximum latitude:", df["latitude"].max())
print("Minimum longitude:", df["longitude"].min())
```

```python
#step 14:check and validate date column
print(df["date_added"].head(10))
print("\nData type:")
print(df["date_added"].dtype)
print("\nMissing values:")
print(df["date_added"].isnull().sum())
#change date to datetime format
df["date_added"]=pd.to_datetime(df["date_added"],format="%m-%d-%Y")
print(df["date_added"].head(10))
print("\nData type:", df["date_added"].dtype)
print("\nMissing after conversion:", df["date_added"].isnull().sum())
```

```python
#step 15: extract date features
df["year"] = df["date_added"].dt.year
df["month"] = df["date_added"].dt.month
df["day"] = df["date_added"].dt.day
print(df[["date_added", "year", "month", "day"]].head(10))
```

```python
#step 16:final check for missing values
print(df.isnull().sum())
```

```python
#step 17: check missing values of agency and agent columns
print("Unique agencies:", df["agency"].nunique())
print("Unique agents:", df["agent"].nunique())
print("\nMissing agency:", df["agency"].isnull().sum())
print("Missing agent:", df["agent"].isnull().sum())
# drop these both columns because these both have 31k+ missing values
df = df.drop(columns=["agency", "agent"])
```

```python
#step 18:final check
print(df.columns)
print("\nShape:", df.shape)
df = df.drop(columns=[
    "property_id",
    "location_id",
    "page_url",
    "purpose",
    "area",
    "date_added"
])
print(df.columns)
print("\nShape:", df.shape)
#save the cleaned dataset to a new CSV file
df.to_csv(r"D:\Summer\House Price Prediction\data\cleaned_house_price_data.csv", index=False)
print("Cleaned dataset saved successfully!")
print(df.shape)
```
